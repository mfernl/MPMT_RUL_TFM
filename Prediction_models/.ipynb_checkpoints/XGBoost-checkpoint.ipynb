```python
import numpy as np
import pandas as pd
from scipy import stats
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error
import xgboost as xgb
from sklearn.model_selection import GroupShuffleSplit
import matplotlib.pyplot as plt
from CMAPSSDataset import *
import CMAPSSDataset as cmap
import time

print(f"XGBoost version: {xgb.__version__}")
```

### Construcción de features por ventana

```python
def construir_features_xgb(df, sensor_cols, window_size, ablation_study):
    """
    Para cada fila (ciclo) construye un vector de features que resume
    la ventana de los últimos window_size ciclos.

    Features por sensor:
      - Valor actual
      - Delta acumulado desde ciclo 1
      - Slope local
      - Aceleración
      - Media, std, min, max, rango sobre la ventana
      - Slope de regresión lineal sobre la ventana completa
      - Último valor menos primer valor de la ventana (tendencia corta)

    Features globales:
      - Ciclo actual (normalizado por motor)
      - Condición operacional
    """
    registros = []

    for motor_id, grupo in df.groupby("motor_id"):
        grupo = grupo.reset_index(drop=True)
        n_ciclos  = len(grupo)
        max_ciclo = grupo["ciclo"].max()

        for t in range(n_ciclos):
            inicio  = max(0, t - window_size + 1)
            ventana = grupo.iloc[inicio:t + 1]

            fila = {}

            # ── Features globales del ciclo ──────────────────────────────
            fila["condicion"] = grupo.loc[t, "condicion"] if "condicion" in grupo.columns else 0

            # ── Configuración operacional actual ─────────────────────────
            for op in ["op_1", "op_2", "op_3"]:
                if op in grupo.columns:
                    fila[op] = grupo.loc[t, op]

            # ── Features por sensor ──────────────────────────────────────
            for col in sensor_cols:
                v = ventana[col].values

                if ablation_study not in [1,3]:
                    # Valor actual y derivadas en t
                    fila[f"{col}_actual"]  = grupo.loc[t, col]
                    fila[f"{col}_delta"]   = grupo.loc[t, f"{col}_delta"] \
                                             if f"{col}_delta" in grupo.columns \
                                             else v[-1] - v[0]
                    fila[f"{col}_slope"]   = grupo.loc[t, f"{col}_slope"] \
                                             if f"{col}_slope" in grupo.columns else 0
                    fila[f"{col}_accel"]   = grupo.loc[t, f"{col}_accel"] \
                                             if f"{col}_accel" in grupo.columns else 0
                if ablation_study not in [2,3]:
                    # Estadísticas sobre la ventana
                    fila[f"{col}_mean"]    = v.mean()
                    fila[f"{col}_std"]     = v.std() if len(v) > 1 else 0
                    fila[f"{col}_min"]     = v.min()
                    fila[f"{col}_max"]     = v.max()
                    fila[f"{col}_rango"]   = v.max() - v.min()
                    # Diferencia entre último y primer ciclo de la ventana
                    fila[f"{col}_delta_ventana"] = v[-1] - v[0]

            fila["rul"]      = grupo.loc[t, "rul"]
            fila["motor_id"] = motor_id
            registros.append(fila)

    return pd.DataFrame(registros)
```

### Preparar Dataset

```python
def preparar_dataset_xgb(ruta_base, subset, ablation_study, window_size=60):
    config        = cmap.CONFIG_DATASETS[subset]
    n_condiciones = config["n_condiciones"]
    rul_max       = config["rul_max"]

    COLS = (
        ["motor_id", "ciclo"] +
        [f"op_{i}" for i in range(1, 4)] +
        [f"s{i}"   for i in range(1, 22)]
    )
    train = pd.read_csv(f"{ruta_base}/train_{subset}.txt", sep=r"\s+", header=None, names=COLS)
    test  = pd.read_csv(f"{ruta_base}/test_{subset}.txt",  sep=r"\s+", header=None, names=COLS)
    rul   = pd.read_csv(f"{ruta_base}/RUL_{subset}.txt",   sep=r"\s+", header=None, names=["rul_final"])

    train_df = cmap.add_rul_train(train.copy(), rul_max=rul_max)
    test_df  = cmap.add_rul_test(test.copy(),   rul, rul_max=rul_max)

    # 1. Eliminar constantes (sobre train)
    sensor_cols = [c for c in train_df.columns if c.startswith("s")]
    std         = train_df[sensor_cols].std()
    constantes  = std[std < 0.01].index.tolist()
    train_df    = train_df.drop(columns=constantes)
    test_df     = test_df.drop(columns=constantes)
    sensor_cols = [c for c in sensor_cols if c not in constantes]

    # 2. SPLIT POR MOTOR — antes de cualquier fit
    motores = train_df["motor_id"].unique()
    gss     = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
    idx_tr, idx_val = next(gss.split(motores, groups=motores))
    motores_train = motores[idx_tr]
    motores_val   = motores[idx_val]

    train_solo = train_df[train_df["motor_id"].isin(motores_train)].copy()
    val_solo   = train_df[train_df["motor_id"].isin(motores_val)].copy()

    print(f"Motores únicos en train: {len(motores_train)}")
    print(f"Motores únicos en val:   {len(motores_val)}")

    # 3. KMeans — fit solo sobre train_solo
    if n_condiciones > 1:
        train_solo, km = cmap.identificar_condicion_operacional(train_solo, n_condiciones)
        val_solo["condicion"]  = km.predict(val_solo[["op_1","op_2","op_3"]])
        test_df["condicion"]   = km.predict(test_df[["op_1","op_2","op_3"]])
    else:
        train_solo["condicion"] = 0
        val_solo["condicion"]   = 0
        test_df["condicion"]    = 0

    # 4. Features de degradación
    print(f"[{subset}] Calculando features de degradación...")
    deg_cols   = [c for c in train_solo.columns if c.startswith("s")]
    train_solo = cmap.añadir_features_degradacion_rapido(train_solo, deg_cols)
    val_solo   = cmap.añadir_features_degradacion_rapido(val_solo,   deg_cols)
    test_df    = cmap.añadir_features_degradacion_rapido(test_df,    deg_cols)

    # 5. Normalización — fit en train_solo, transform en val y test
    feature_cols = [c for c in train_solo.columns if c.startswith("s") or c.startswith("op")]
    train_solo, val_solo, test_df, scalers = cmap.normalizar_por_condicion(
        train_solo, val_solo, test_df, feature_cols, n_condiciones
    )

    # 6. Tabularización — features agregadas por ventana
    print(f"[{subset}] Construyendo features agregadas...")
    train_feat = construir_features_xgb(train_solo, sensor_cols, window_size, ablation_study)
    val_feat   = construir_features_xgb(val_solo, sensor_cols, window_size, ablation_study)
    test_feat  = construir_features_xgb(test_df, sensor_cols, window_size, ablation_study)

    # Test: solo el último ciclo de cada motor
    test_feat = test_feat.groupby("motor_id").last().reset_index()

    xgb_feature_cols = [c for c in train_feat.columns
                        if c not in ["rul", "motor_id", "ciclo_actual"]]

    X_train = train_feat[xgb_feature_cols].values.astype(np.float32)
    y_train = train_feat["rul"].values.astype(np.float32)
    X_val   = val_feat[xgb_feature_cols].values.astype(np.float32)
    y_val   = val_feat["rul"].values.astype(np.float32)
    X_test  = test_feat[xgb_feature_cols].values.astype(np.float32)
    y_test  = test_feat["rul"].values.astype(np.float32)

    print(f"[{subset}] Train {X_train.shape} | Val {X_val.shape} | Test {X_test.shape}")
    print(f"[{subset}] Features XGB: {len(xgb_feature_cols)} | Condiciones: {n_condiciones}")

    return (X_train, y_train), (X_val, y_val), (X_test, y_test), scalers, xgb_feature_cols
```

### Entrenamiento XGBoost

```python
def nasa_score_numpy(y_true, y_pred):
    diff = y_pred - y_true
    return np.sum(np.where(diff < 0,
                           np.exp(-diff / 13.0) - 1,
                           np.exp( diff / 10.0) - 1))

def entrenar_xgb_completo(subset, ruta_base, ablation_study, window_size=60,learning_rate=0.05, early_stopping=50,n_ensemble=10):
    """
    Entrena:
      - 1 modelo de regresión puntual (igual que antes)
      - 3 modelos de cuantiles (p10, p50, p90) — incertidumbre aleatoria
      - n_ensemble modelos con semillas distintas — incertidumbre epistémica
    """
    cfg = cmap.CONFIG_DATASETS[subset]
    print(f"Parámetros de {subset}: rul_max = {cfg['rul_max']} | "
          f"condiciones = {cfg['n_condiciones']}")

    # ── Datos ────────────────────────────────────────────────────────
    t0 = time.time()
    (X_tr, y_tr), (X_val, y_val), (X_te, y_te), scalers, feature_cols = \
        preparar_dataset_xgb(ruta_base, subset, ablation_study, window_size)
    t_datos = time.time() - t0

    w_tr = np.where(y_tr < 30, 4.0, np.where(y_tr < 80, 2.0, 1.0))
    
    def base_params(extra={}):
        p = dict(
            n_estimators          = 3000,
            learning_rate         = learning_rate,
            max_depth             = 6,
            min_child_weight      = 5,
            subsample             = 0.8,
            colsample_bytree      = 0.8,
            gamma                 = 0.1,
            reg_alpha             = 0.1,
            reg_lambda            = 1.0,
            early_stopping_rounds = early_stopping,
            device                = "cuda",
        )
        p.update(extra)
        return p

    # Kwargs comunes — para modelo puntual y ensemble
    fit_kwargs = dict(
        sample_weight = w_tr,
        eval_set      = [(X_val, y_val)],
        verbose       = False
    )
    
    # ── 1. Modelo puntual (igual que antes) ──────────────────────────
    print(f"\n[{subset}] Entrenando modelo puntual...")
    t0 = time.time()
    model_point = xgb.XGBRegressor(
        **base_params({"objective": "reg:squarederror",
                       "eval_metric": "rmse",
                       "random_state": 42})
    )
    model_point.fit(X_tr, y_tr, **fit_kwargs)
    print(f"  Árboles: {model_point.best_iteration + 1} | "
          f"Tiempo: {(time.time()-t0)/60:.2f} min")

    # ── 2. Cuantiles p10, p50, p90 — incertidumbre aleatoria ─────────
    print(f"[{subset}] Entrenando modelos de cuantiles...")
    models_quantile = {}
    for alpha in [0.1, 0.5, 0.9]:
        t0 = time.time()
        
        # p90 no se beneficia de early stopping — entrena árboles fijos
        n_trees = 500 if alpha == 0.9 else 3000
        stopping = None if alpha == 0.9 else early_stopping
        
        m = xgb.XGBRegressor(
            n_estimators          = n_trees,
            learning_rate         = learning_rate,
            max_depth             = 6,
            min_child_weight      = 5,
            subsample             = 0.8,
            colsample_bytree      = 0.8,
            gamma                 = 0.1,
            reg_alpha             = 0.1,
            reg_lambda            = 1.0,
            early_stopping_rounds = stopping,
            objective             = "reg:quantileerror",
            quantile_alpha        = alpha,
            eval_metric           = "quantile",
            device                = "cuda",
            random_state          = 42,
        )
        
        fit_q = dict(eval_set=[(X_val, y_val)], verbose=False)
        m.fit(X_tr, y_tr, **fit_q)
        models_quantile[alpha] = m
        print(f"  p{int(alpha*100):02d} | Árboles: {n_trees if alpha == 0.9 else m.best_iteration + 1} | "
              f"Tiempo: {(time.time()-t0)/60:.2f} min")

    # ── 3. Deep ensembles — incertidumbre epistémica ──────────────────
    print(f"[{subset}] Entrenando ensemble ({n_ensemble} modelos)...")
    models_ensemble = []
    t0 = time.time()
    for seed in range(n_ensemble):
        m = xgb.XGBRegressor(
            **base_params({
                "objective":         "reg:squarederror",
                "eval_metric":       "rmse",
                "random_state":       seed,
                "subsample":          0.6,    # antes 0.8 — más diversidad
                "colsample_bytree":   0.6,    # antes 0.8 — más diversidad
                "max_depth":          np.random.randint(4, 8),  # profundidad aleatoria
                "learning_rate":      learning_rate * np.random.uniform(0.8, 1.2),
            })
        )
        m.fit(X_tr, y_tr, **fit_kwargs)
        models_ensemble.append(m)

    return {
        "point":    model_point,
        "quantile": models_quantile,
        "ensemble": models_ensemble,
        "test":     (X_te, y_te),
        "features": feature_cols,
    }


def evaluar_xgb_completo(resultado, subset):
    X_te, y_te = resultado["test"]

    # ── Predicción puntual ───────────────────────────────────────────
    y_pred = resultado["point"].predict(X_te)
    diff   = y_pred - y_te
    rmse_v = np.sqrt(np.mean(diff**2))
    mae_v  = np.mean(np.abs(diff))
    score  = np.where(diff < 0,
                      np.exp(-diff/13.0)-1,
                      np.exp( diff/10.0)-1).sum()

    # ── Cuantiles — incertidumbre aleatoria ──────────────────────────
    p10 = resultado["quantile"][0.1].predict(X_te)
    p50 = resultado["quantile"][0.5].predict(X_te)
    p90 = resultado["quantile"][0.9].predict(X_te)

    amplitud_q  = (p90 - p10).mean()
    cobertura_q = np.mean((y_te >= p10) & (y_te <= p90)) * 100

    # ── Ensemble — incertidumbre epistémica ──────────────────────────
    ens_preds = np.stack([m.predict(X_te)
                          for m in resultado["ensemble"]])  # (n, motores)
    ens_media = ens_preds.mean(axis=0)
    ens_std   = ens_preds.std(axis=0)

    lower_e     = ens_media - 1.96 * ens_std
    upper_e     = ens_media + 1.96 * ens_std
    cobertura_e = np.mean((y_te >= lower_e) & (y_te <= upper_e)) * 100

    print(f"\n── XGBoost {subset} ─────────────────────────────────────")
    print(f"RMSE puntual               : {rmse_v:.2f}")
    print(f"MAE                        : {mae_v:.2f}")
    print(f"Score                      : {score:.1f}")

    print(f"\n── Incertidumbre aleatoria (cuantiles) ──────────────────")
    print(f"Amplitud media [p10, p90]  : {amplitud_q:.2f} ciclos")
    print(f"Cobertura empírica 80%     : {cobertura_q:.1f}%  (objetivo: 80%)")

    print(f"\n── Incertidumbre epistémica (ensemble) ──────────────────")
    print(f"Std media ensemble         : {ens_std.mean():.2f} ciclos")
    print(f"Cobertura empírica 95%     : {cobertura_e:.1f}%  (objetivo: 95%)")

    return {
        "RMSE puntual": rmse_v,
        "Score": score
    }
```

### Evaluar XGB

```python
RUTA = "../CMAPSSData"
resultados_xgb = {}

for subset in ["FD001", "FD002", "FD003", "FD004"]:
    print(f"\n{'='*50}\n{subset}")
    resultado = entrenar_xgb_completo(subset, RUTA, n_ensemble=10)
    r = evaluar_xgb_completo(resultado, subset)
    resultados_xgb[subset] = r
```

## ABLATION STUDY

```python
RUTA = "../CMAPSSData"
resultados_xgb = {}
resultados_ablation = {}
ablation = ["completo","sin_features_derivadas","sin_estadísticos_ventana","raw_data"]
for i in ablation:
    for subset in ["FD001", "FD002", "FD003", "FD004"]:
        print(f"\n{'='*20} ABLATION:{i}{'='*20}\n{subset}")
        resultado = entrenar_xgb_completo(subset, RUTA, ablation_study=ablation.index(i), n_ensemble=10)
        r = evaluar_xgb_completo(resultado, subset)
        resultados_xgb[subset] = r
    resultados_ablation[i] = resultados_xgb.copy()
print(resultados_ablation)
```

```python
## Ejemplo de uso real de predicción con incertidumbre
def predecir_motor(idx_motor, resultado, subset):
    X_te, y_te = resultado["test"]

    X_motor = X_te[idx_motor:idx_motor+1]
    y_real  = y_te[idx_motor]

    y_punto = resultado["point"].predict(X_motor)[0]
    p10     = resultado["quantile"][0.1].predict(X_motor)[0]
    p50     = resultado["quantile"][0.5].predict(X_motor)[0]
    p90     = resultado["quantile"][0.9].predict(X_motor)[0]

    ens_preds = np.array([m.predict(X_motor)[0]
                          for m in resultado["ensemble"]])
    ens_media = ens_preds.mean()
    ens_std   = ens_preds.std()

    dentro_q = p10 <= y_real <= p90
    dentro_e = (ens_media - 1.96*ens_std) <= y_real <= (ens_media + 1.96*ens_std)

    print(f"\nMotor {idx_motor + 1} — {subset}")
    print(f"RUL real                   : {y_real:.0f} ciclos")
    print(f"Predicción puntual         : {y_punto:.1f} ciclos")
    print(f"\nIncertidumbre aleatoria (cuantiles):")
    print(f"  p10 / p50 / p90          : {p10:.1f} / {p50:.1f} / {p90:.1f}")
    print(f"  Amplitud [p10, p90]      : {p90-p10:.1f} ciclos")
    print(f"  RUL real en [p10, p90]   : {'Sí' if dentro_q else 'No'}")
    print(f"\nIncertidumbre epistémica (ensemble):")
    print(f"  Media ensemble           : {ens_media:.1f} ciclos")
    print(f"  Std ensemble             : {ens_std:.1f} ciclos")
    print(f"  IC 95% ensemble          : [{ens_media-1.96*ens_std:.1f}, {ens_media+1.96*ens_std:.1f}]")
    print(f"  RUL real en IC 95%       : {'Sí' if dentro_e else 'No'}")
```

```python
predecir_motor(idx_motor=0, resultado=resultados_xgb["FD001"]["modelos"], subset="FD001")
```

### Interpretación de features

```python
def plot_feature_importance(model, feature_cols, subset, top_n=20):
    importancias = model.feature_importances_
    idx          = np.argsort(importancias)[::-1][:top_n]

    plt.figure(figsize=(6, 6))
    plt.barh(range(top_n),
             importancias[idx][::-1],
             color="steelblue", alpha=0.8)
    plt.yticks(range(top_n),
               [feature_cols[i] for i in idx][::-1],
               fontsize=9)
    plt.xlabel("Importancia")
    plt.title(f"{subset} — Top {top_n} features más predictivas")
    plt.tight_layout()
    plt.show()

    print(f"\nTop 10 features — {subset}:")
    for i in idx[:10]:
        print(f"  {feature_cols[i]:35s}  {importancias[i]:.4f}")

for subset, res in resultados_xgb.items():
    plot_feature_importance(
        res["model"], res["feats"], subset, top_n=20)
```
