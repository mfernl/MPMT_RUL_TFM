```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from scipy import stats
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import GroupShuffleSplit
from CMAPSSDataset import *
import CMAPSSDataset as cmap
import time

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Usando: {DEVICE}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"VRAM disponible: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
```

### Construcción del modelo

```python
# ── Atención Bahdanau ────────────────────────────────────────────────────────

class BahdanauAttention(nn.Module):
    def __init__(self, hidden_dim, attn_units=64):
        super().__init__()
        self.W = nn.Linear(hidden_dim, attn_units)
        self.v = nn.Linear(attn_units, 1, bias=False)

    def forward(self, h):
        # h: (batch, T, hidden_dim)
        score   = self.v(torch.tanh(self.W(h)))        # (batch, T, 1)
        weights = torch.softmax(score, dim=1)           # (batch, T, 1)
        context = torch.sum(weights * h, dim=1)         # (batch, hidden_dim)
        return context, weights


# ── Modelo completo ──────────────────────────────────────────────────────────

class CNNBiLSTMAttention(nn.Module):
    def __init__(self, n_features, window_size=30,
                 cnn_filtros=64, lstm_units=64,
                 attn_units=64, dropout=0.3):
        super().__init__()

        # ── CNN 1D ──────────────────────────────────────────────────────────
        # En PyTorch Conv1d espera (batch, channels, length)
        # Entrada: (batch, window_size, n_features) → permutar antes de Conv1d
        self.cnn = nn.Sequential(
            nn.Conv1d(n_features, cnn_filtros, kernel_size=3, padding=1),
            nn.BatchNorm1d(cnn_filtros),
            nn.ReLU(),
            nn.Dropout(), # MCD implementacion
            nn.Conv1d(cnn_filtros, cnn_filtros, kernel_size=5, padding=2),
            nn.BatchNorm1d(cnn_filtros),
            nn.ReLU(),
            nn.Dropout(), # MCD implementacion
            nn.Conv1d(cnn_filtros, cnn_filtros, kernel_size=3,
                      padding=2, dilation=2),
            nn.BatchNorm1d(cnn_filtros),
            nn.ReLU(),
            nn.Dropout(), # MCD implementacion
        )

        # ── BiLSTM ──────────────────────────────────────────────────────────
        # Entrada tras CNN: (batch, window_size, cnn_filtros)
        self.bilstm1 = nn.LSTM(
            input_size   = cnn_filtros,
            hidden_size  = lstm_units,
            batch_first  = True,
            bidirectional= True,
        )
        
        self.bilstm2 = nn.LSTM(
            input_size   = lstm_units * 2,   # *2 por bidireccional
            hidden_size  = lstm_units,
            batch_first  = True,
            bidirectional= True
        )
        self.dropout = nn.Dropout(dropout)

        # ── Atención ────────────────────────────────────────────────────────
        self.attention = BahdanauAttention(
            hidden_dim = lstm_units * 2,
            attn_units = attn_units
        )

        """
        # ── Cabeza de regresión ──────────────────────────────────────────────
        self.fc = nn.Sequential(
            nn.Linear(lstm_units * 2, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(dropout), # generalizar
            nn.Linear(64, 1)
        )
        """

        # ── Tronco compartido ────────────────────────────────────────
        self.fc_shared = nn.Sequential(
            nn.Linear(lstm_units * 2, 128),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(dropout),
        )

        # ── Cabeza media (mu) ────────────────────────────────────────
        self.fc_mu = nn.Linear(64, 1)

        # ── Cabeza log-varianza (log_var) ────────────────────────────
        # Predice log(σ²) — mantenerlo en log evita valores negativos
        self.fc_log_var = nn.Linear(64, 1)

    def forward(self, x):
        # x: (batch, window_size, n_features)

        # CNN — necesita (batch, channels, length)
        x_cnn = x.permute(0, 2, 1)          # (batch, n_features, window_size)
        x_cnn = self.cnn(x_cnn)             # (batch, cnn_filtros, window_size)
        x_cnn = x_cnn.permute(0, 2, 1)      # (batch, window_size, cnn_filtros)

        # BiLSTM
        x_lstm, _ = self.bilstm1(x_cnn)     # (batch, window_size, lstm*2)
        x_lstm = self.dropout(x_lstm)
        x_lstm, _ = self.bilstm2(x_lstm)    # (batch, window_size, lstm*2)
        x_lstm = self.dropout(x_lstm)

        # Atención
        context, attn_w = self.attention(x_lstm)   # (batch, lstm*2)

        # Regresión
        #out = self.fc(context)               # (batch, 1)
        
        shared  = self.fc_shared(context)
        mu      = self.fc_mu(shared).squeeze(-1)       # (batch,)
        log_var = self.fc_log_var(shared).squeeze(-1)  # (batch,)
        return mu, log_var, attn_w
        
        #return out.squeeze(-1), attn_w       # (batch,), (batch, T, 1)
```

### Preparar dataset

```python
def preparar_dataset(ruta_base,subset,window_size=60):

    config = cmap.CONFIG_DATASETS[subset]
    n_condiciones = config["n_condiciones"]
    rul_max = config["rul_max"]

    COLS = (
        ["motor_id", "ciclo"] +
        [f"op_{i}" for i in range(1, 4)] +
        [f"s{i}"   for i in range(1, 22)]
    )
    train = pd.read_csv(f"{ruta_base}/train_{subset}.txt", sep=r"\s+", header=None, names=COLS)
    test  = pd.read_csv(f"{ruta_base}/test_{subset}.txt",  sep=r"\s+", header=None, names=COLS)
    rul = pd.read_csv(f"{ruta_base}/RUL_{subset}.txt",   sep=r"\s+", header=None, names=["rul_final"])

    train_df = cmap.add_rul_train(train.copy(), rul_max=rul_max)
    test_df  = cmap.add_rul_test(test.copy(),   rul, rul_max=rul_max)

    # 1. Eliminar constantes (sobre train)
    sensor_cols = [c for c in train_df.columns if c.startswith("s")]
    std         = train_df[sensor_cols].std()
    constantes  = std[std < 0.01].index.tolist()
    train_df    = train_df.drop(columns=constantes)
    test_df     = test_df.drop(columns=constantes)
    sensor_cols = [c for c in sensor_cols if c not in constantes]

    # 2. SPLIT POR MOTOR — antes de cualquier fit
    motores = train_df["motor_id"].unique()
    gss     = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
    idx_tr, idx_val = next(gss.split(motores, groups=motores))
    motores_train = motores[idx_tr]
    motores_val   = motores[idx_val]

    train_solo = train_df[train_df["motor_id"].isin(motores_train)].copy()
    val_solo   = train_df[train_df["motor_id"].isin(motores_val)].copy()

    print(f"Motores únicos en train: {len(motores_train)}")
    print(f"Motores únicos en val:   {len(motores_val)}")

    # 3. KMeans — fit solo sobre train_solo
    if n_condiciones > 1:
        train_solo, km = cmap.identificar_condicion_operacional(train_solo, n_condiciones)
        val_solo["condicion"] = km.predict(val_solo[["op_1","op_2","op_3"]])
        test_df["condicion"]  = km.predict(test_df[["op_1","op_2","op_3"]])
    else:
        train_solo["condicion"] = 0
        val_solo["condicion"]   = 0
        test_df["condicion"]    = 0

    # 4. Features de degradación
    print(f"[{subset}] Calculando features de degradación...")
    deg_cols   = [c for c in train_solo.columns if c.startswith("s")]
    train_solo = cmap.añadir_features_degradacion_rapido(train_solo, deg_cols)
    val_solo   = cmap.añadir_features_degradacion_rapido(val_solo,   deg_cols)
    test_df    = cmap.añadir_features_degradacion_rapido(test_df,    deg_cols)

    # 5. Normalización — fit en train_solo, transform en val y test
    feature_cols = [c for c in train_solo.columns if c.startswith("s") or c.startswith("op")]
    train_solo, val_solo, test_df, scalers = cmap.normalizar_por_condicion(
        train_solo, val_solo, test_df, feature_cols, n_condiciones
    )

    # 6. Ventanas
    X_train, y_train = cmap.crear_ventanas(train_solo, feature_cols, window_size)
    X_val,   y_val   = cmap.crear_ventanas(val_solo,   feature_cols, window_size)
    X_test,  y_test  = cmap.crear_ventanas_test(test_df, feature_cols, window_size)

    print(f"[{subset}] Train: {X_train.shape} | Val: {X_val.shape} | Test: {X_test.shape}")
    print(f"[{subset}] Features: {len(feature_cols)} | Condiciones: {n_condiciones}")

    return (X_train, y_train), (X_val, y_val), (X_test, y_test), scalers, feature_cols
```

## Entrenar y evaular el dataset

```python
def nasa_score(diff):
    score_por_motor = np.where(
        diff < 0,
        np.exp(-diff / 13.0) - 1,
        np.exp( diff / 10.0) - 1
    )
    return score_por_motor

def loss_ponderada_rmse(y_pred, y_true):
    diff = y_pred - y_true
    w = torch.where(y_true < 30,
            torch.full_like(y_true, 4.0),
            torch.where(y_true < 80,
                        torch.full_like(y_true, 2.0),
                        torch.ones_like(y_true)))
    return torch.sqrt((w * diff**2).mean())

def nll_ponderada(mu, log_var, y_true):
    """
    Negative log-likelihood gaussiana con pesos por ciclos críticos.
    
    El modelo aprende simultáneamente a predecir la media (mu) y la
    varianza (exp(log_var)) de la distribución del error. En ciclos
    críticos (RUL bajo) se penaliza más el error en la media.
    """
    var = torch.exp(log_var) + 1e-6  # varianza siempre positiva

    # Pesos por ciclos críticos — igual que antes
    w = torch.where(y_true < 30,
            torch.full_like(y_true, 4.0),
            torch.where(y_true < 80,
                        torch.full_like(y_true, 2.0),
                        torch.ones_like(y_true)))

    # NLL gaussiana: 0.5 * (log_var + (y - mu)² / var)
    # Los pesos se aplican al término de error cuadrático
    nll = 0.5 * (log_var + w * (y_true - mu)**2 / var)
    return nll.mean()
```

```python
def entrenar(subset, ruta_base="../CMAPSSData",
             window_size=60, epochs=150, batch_size=256):

    cfg = cmap.CONFIG_DATASETS[subset]
    cnn_filtros = cfg["cnn_filtros"]
    lstm_units = cfg["lstm_units"]
    print(f"Parámetros de {subset}: cnn_filtros = {cnn_filtros}, lstm_units = {lstm_units}")
    
    # Datos
    t0 = time.time()
    (X_tr, y_tr), (X_val, y_val), (X_te, y_te), scalers, feats = \
        preparar_dataset(ruta_base, subset, window_size)
    t_datos = time.time() - t0

    train_loader = DataLoader(
        cmap.CMAPSSDataset(X_tr, y_tr),
        batch_size=batch_size, shuffle=True,
        pin_memory=True,        # acelera transferencia CPU→GPU
        num_workers=4
    )
    val_loader = DataLoader(
        cmap.CMAPSSDataset(X_val, y_val),
        batch_size=batch_size, shuffle=False,
        pin_memory=True
    )

    # Modelo
    model = CNNBiLSTMAttention(
        n_features  = X_tr.shape[2],
        window_size = window_size,
        cnn_filtros = cnn_filtros,
        lstm_units  = lstm_units,
        attn_units  = 64,
        dropout     = 0.3
    ).to(DEVICE)

    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay = 1e-4) #antes sin weight_decay
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, factor=0.5, patience=10, min_lr=1e-6)

    mejor_val_rmse = float("inf")
    paciencia_actual = 0
    PACIENCIA_MAX = 50
    historial = {"train_rmse": [], "val_rmse": []}
    t0 = time.time()
    for epoch in range(1, epochs + 1):

        # ── Entrenamiento ────────────────────────────────────────────────
        model.train()
        train_loss = 0.0

        for X_batch, y_batch in train_loader:
            X_batch = X_batch.to(DEVICE)
            y_batch = y_batch.to(DEVICE)

            optimizer.zero_grad()
            mu, log_var, _ = model(X_batch)          # CAMBIO: 3 salidas
            #y_pred, _ = model(X_batch)
            
            #loss = loss_ponderada_rmse(y_pred, y_batch)  
            loss = nll_ponderada(mu, log_var, y_batch)  # CAMBIO: nueva pérdida

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()
            train_loss += loss.item()

        train_loss /= len(train_loader)

        # ── Validación ───────────────────────────────────────────────────
        model.eval()
        val_preds_list, val_trues_list = [], []

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch = X_batch.to(DEVICE)
                y_batch = y_batch.to(DEVICE)
                #y_pred, _ = model(X_batch)
                #val_preds_list.append(y_pred.cpu())
                mu, log_var, _ = model(X_batch)      # CAMBIO: 3 salidas
                val_preds_list.append(mu.cpu())       # CAMBIO: solo mu para RMSE
                val_trues_list.append(y_batch.cpu())

        val_preds_all = torch.cat(val_preds_list)
        val_trues_all = torch.cat(val_trues_list)
        val_loss = torch.sqrt(torch.mean((val_preds_all - val_trues_all) ** 2)).item()

        scheduler.step(val_loss)
        historial["train_rmse"].append(train_loss)
        historial["val_rmse"].append(val_loss)

        if epoch % 10 == 0:
            lr_actual = optimizer.param_groups[0]["lr"]
            print(f"Época {epoch:>3} | train_as_loss {train_loss:.4f} | "
                  f"val_rmse {val_loss:.4f} | lr {lr_actual:.2e}")

        # ── Early stopping y checkpoint ──────────────────────────────────
        if val_loss < mejor_val_rmse:
            mejor_val_rmse = val_loss
            paciencia_actual = 0
            torch.save(model.state_dict(), f"mejor_modelo_{subset}.pt")
        else:
            paciencia_actual += 1
            if paciencia_actual >= PACIENCIA_MAX:
                print(f"Early stopping en época {epoch}")
                break

    # Cargar mejor checkpoint
    t_train = (time.time() - t0) / 60
    print(f"[{subset}] Tiempo preprocesando: {t_datos/60}")
    print(f"[{subset}] Tiempo entrenando: {t_train}")
    model.load_state_dict(torch.load(f"mejor_modelo_{subset}.pt"))
    return model, historial, (X_te, y_te)


def evaluar(model, X_test, y_test, subset, n_passes=50):
    activar_mc_dropout(model)
    X_tensor = torch.tensor(X_test, dtype=torch.float32).to(DEVICE)

    mus, log_vars = [], []
    with torch.no_grad():
        for _ in range(n_passes):
            mu, log_var, attn_w = model(X_tensor)
            mus.append(mu.cpu().numpy())
            log_vars.append(log_var.cpu().numpy())

    mus = np.stack(mus)       # (n_passes, n_motores)
    log_vars = np.stack(log_vars)  # (n_passes, n_motores)

    # Predicción puntual
    y_pred = mus.mean(axis=0)

    # Incertidumbre epistémica — desacuerdo entre pasadas MCD
    std_ep = mus.std(axis=0)
    lower_ep = y_pred - 1.96 * std_ep
    upper_ep = y_pred + 1.96 * std_ep
    cobertura_ep = np.mean((y_test >= lower_ep) & (y_test <= upper_ep)) * 100

    # Incertidumbre aleatoria — varianza predicha por la cabeza NLL
    std_al = np.sqrt(np.exp(log_vars).mean(axis=0))
    lower_al = y_pred - 1.96 * std_al
    upper_al = y_pred + 1.96 * std_al
    cobertura_al = np.mean((y_test >= lower_al) & (y_test <= upper_al)) * 100

    diff   = y_pred - y_test
    rmse_v = np.sqrt(np.mean(diff**2))
    mae_v  = np.mean(np.abs(diff))
    score  = nasa_score(diff).sum()

    print(f"% predicciones tardías : {(diff > 0).mean()*100:.1f}%")
    print(f"Error medio tardío     : {diff[diff > 0].mean():.1f}")
    print(f"Error medio temprano   : {diff[diff < 0].mean():.1f}")

    print(f"\n── {subset} ──────────────────────────")
    print(f"RMSE                   : {rmse_v:.2f}")
    print(f"MAE                    : {mae_v:.2f}")
    print(f"Score                  : {score:.1f}")

    print(f"\n── Incertidumbre epistémica (MCD) {subset} ──")
    print(f"Std media              : {std_ep.mean():.2f} ciclos")
    print(f"Cobertura IC 95%       : {cobertura_ep:.1f}%  (objetivo: 95%)")

    print(f"\n── Incertidumbre aleatoria (NLL) {subset} ───")
    print(f"Std media              : {std_al.mean():.2f} ciclos")
    print(f"Cobertura IC 95%       : {cobertura_al:.1f}%  (objetivo: 95%)")

    return y_pred, std_ep, std_al, attn_w.cpu().numpy()
```

```python
def activar_mc_dropout(model):
    """
    Activa dropout manteniendo BatchNorm en modo eval.
    """
    model.eval()  # primero todo en eval (congela BN)
    for m in model.modules():
        if isinstance(m, torch.nn.Dropout):
            m.train()  # solo dropout en modo train
```

```python
# Train
resultados = {}

for subset in ["FD001", "FD002", "FD003", "FD004"]:
    print(f"\n{'='*50}\nEntrenando {subset}...")
    model, hist, (X_te, y_te) = entrenar(subset)
    y_pred, std_ep, std_al, attn_w = evaluar(model, X_te, y_te, subset)
    resultados[subset] = {
        "y_pred"  : y_pred,
        "y_te"    : y_te,
        "std_ep"  : std_ep,
        "std_al"  : std_al,
        "hist"    : hist,
    }
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

def evaluar_kmeans(ruta_base, subset, max_k=10):
    train_raw = pd.read_csv(f"../{ruta_base}/train_{subset}.txt", sep=r"\s+", header=None, names=COLS)
    op = train_raw[["op_1", "op_2", "op_3"]].values

    inercias        = []
    silhouettes     = []
    davies_bouldins = []
    ks = range(2, max_k + 1)

    for k in ks:
        km  = KMeans(n_clusters=k, random_state=42, n_init=10)
        lbl = km.fit_predict(op)
        inercias.append(km.inertia_)
        silhouettes.append(silhouette_score(op, lbl, sample_size=5000))
        davies_bouldins.append(davies_bouldin_score(op, lbl))
        print(f"k={k:>2} | inercia={km.inertia_:>10.1f} | "
              f"silhouette={silhouettes[-1]:.4f} | "
              f"davies-bouldin={davies_bouldins[-1]:.4f}")

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))

    # Elbow — buscar el codo donde la inercia deja de bajar bruscamente
    axes[0].plot(ks, inercias, "o-")
    axes[0].set_xlabel("k"); axes[0].set_ylabel("Inercia")
    axes[0].set_title("Método del codo")
    axes[0].axvline(x=6, color="red", linestyle="--", alpha=0.5, label="k=6")
    axes[0].legend()

    # Silhouette — cuanto más alto mejor (máximo = 1)
    # Mide qué tan bien separados están los clusters
    axes[1].plot(ks, silhouettes, "o-")
    axes[1].set_xlabel("k"); axes[1].set_ylabel("Silhouette score")
    axes[1].set_title("Silhouette (mayor = mejor)")
    axes[1].axvline(x=6, color="red", linestyle="--", alpha=0.5, label="k=6")
    axes[1].legend()

    # Davies-Bouldin — cuanto más bajo mejor (mínimo = 0)
    # Mide la similitud entre clusters — clusters compactos y bien separados → bajo
    axes[2].plot(ks, davies_bouldins, "o-")
    axes[2].set_xlabel("k"); axes[2].set_ylabel("Davies-Bouldin")
    axes[2].set_title("Davies-Bouldin (menor = mejor)")
    axes[2].axvline(x=6, color="red", linestyle="--", alpha=0.5, label="k=6")
    axes[2].legend()

    plt.suptitle(f"{subset} — evaluación número óptimo de clusters")
    plt.tight_layout()
    plt.show()

evaluar_kmeans("CMAPSSData", "FD002")
evaluar_kmeans("CMAPSSData", "FD004")
```

```python
def mostrar_centros_kmeans(ruta_base, subset, k=6):
    train_raw = pd.read_csv(f"../{ruta_base}/train_{subset}.txt", sep=r"\s+", header=None, names=COLS)
    op = pd.DataFrame(train_raw[["op_1", "op_2", "op_3"]].values)
    
    km = KMeans(n_clusters=k, random_state=42)
    clusters = km.fit_predict(op)
    print(f"-------Tamaño de cada cluster ------------")
    for c in np.unique(clusters):
        group = op.iloc[np.where(clusters == c)]
        print(f"cluster {c}: {len(group)}")
    print(f"----Centros de cada cluster: {subset}-----")
    for c in np.unique(clusters):
        group = op.iloc[np.where(clusters == c)]
        centro = group.mean().round(4)
        print(f"op_1: {centro[0]} | op_2: {centro[1]} | op_3: {centro[2]}")
```

```python
mostrar_centros_kmeans("CMAPSSData", "FD002")
```

```python
mostrar_centros_kmeans("CMAPSSData", "FD004")
```

## Justificación RUL MAX

```python
COLS = (
["motor_id", "ciclo"] +
[f"op_{i}" for i in range(1, 4)] +          # 3 configuraciones operacionales
[f"s{i}" for i in range(1, 22)]             # 21 sensores
)
for subset in ["FD001", "FD002", "FD003", "FD004"]:
    train = pd.read_csv(f"../CMAPSSData/train_{subset}.txt",
                        sep=r"\s+", header=None, names=COLS)
    max_ciclos = train.groupby("motor_id")["ciclo"].max()
    print(f"\n{subset}:")
    print(f"  Ciclos por motor — min: {max_ciclos.min():.0f}  "
          f"media: {max_ciclos.mean():.0f}  "
          f"max: {max_ciclos.max():.0f}  "
          f"p75: {max_ciclos.quantile(0.75):.0f}")

for subset, rul_max in [("FD001",125), ("FD002",125), ("FD003",150), ("FD004",150)]:
    train = pd.read_csv(f"../CMAPSSData/train_{subset}.txt",
                        sep=r"\s+", header=None, names=COLS)
    max_ciclo = train.groupby("motor_id")["ciclo"].max()
    train = train.join(max_ciclo.rename("ciclo_max"), on="motor_id")
    train["rul"] = (train["ciclo_max"] - train["ciclo"]).clip(upper=rul_max)

    pct_truncado = (train["rul"] == rul_max).mean() * 100
    print(f"{subset} (rul_max={rul_max}): "
          f"{pct_truncado:.1f}% de ciclos con rul={rul_max} "
          f"(etiqueta constante)")
```
